# Pianist Transformer Colab GPU Inference

Purpose: reproduce pretrained Pianist Transformer inference on one official example MIDI using Google Colab GPU.

This notebook is for **Google Colab GPU runtime**, not for local Windows CPU execution. The local PC currently has no NVIDIA GPU/CUDA path, so the first successful reproduction is planned on Colab.

Run order:

1. Select a Colab GPU runtime: `Runtime > Change runtime type > Hardware accelerator > GPU`.
2. Run the environment check cell and stop if CUDA is unavailable.
3. Define paths in one place.
4. Clone the official repository if needed.
5. Install only missing non-GUI dependencies; do not reinstall PyTorch unless a real incompatibility is confirmed.
6. Import official Pianist Transformer modules.
7. Download the official fine-tuned rendering checkpoint if missing.
8. Select one bundled example MIDI.
9. Load the official model class, move the model to CUDA explicitly, and reuse official generation functions.
10. Run inference, verify output, then download the generated MIDI.

Notes:

- The official `script/inference.sh` is not used because the official `src/inference/inference.py` currently hardcodes `device="cpu"`.
- This notebook does not modify any files inside the official source tree.
- Full model training is out of scope.

```python
import importlib.metadata as metadata
import subprocess
import sys

# Official README recommends Python 3.11, PyTorch 2.7.1, and Transformers 4.54.0.
# Colab often already provides CUDA-enabled PyTorch. We do not reinstall PyTorch automatically,
# because replacing Colab's torch wheel can force a runtime restart or break CUDA compatibility.
EXPECTED_TORCH = "2.7.1"
installed_torch = torch.__version__.split("+")[0]
print("Installed torch:", torch.__version__)
if installed_torch != EXPECTED_TORCH:
    print("WARNING: Official PyTorch version is 2.7.1, but this runtime has", torch.__version__)
    print("TODO: If inference fails with a torch/transformers error, install the official torch build for the active CUDA runtime and restart the runtime.")

# Minimal command-line inference does not need GUI packages PyQt5 and pygame.
# ModelScope is optional because this notebook uses the official Hugging Face checkpoint path.
REQUIRED_PACKAGES = {
    "transformers": "4.54.0",
    "datasets": "4.0.0",
    "accelerate": "1.10.1",
    "miditoolkit": "1.0.1",
    "wandb": "0.21.4",
    "partitura": "1.7.0",
    "seaborn": "0.13.2",
    "huggingface_hub": "0.35.3",
}
OPTIONAL_SKIPPED = {
    "modelscope": "1.32.0",
    "PyQt5": "5.15.11",
    "pygame": "2.6.1",
}

missing_or_mismatched = []
for package_name, expected_version in REQUIRED_PACKAGES.items():
    try:
        installed_version = metadata.version(package_name)
    except metadata.PackageNotFoundError:
        installed_version = None
    print(f"{package_name}: installed={installed_version}, expected={expected_version}")
    if installed_version != expected_version:
        missing_or_mismatched.append(f"{package_name}=={expected_version}")

print("Optional packages skipped for minimal inference:", OPTIONAL_SKIPPED)

if missing_or_mismatched:
    print("Installing missing/mismatched non-GUI dependencies:", missing_or_mismatched)
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", *missing_or_mismatched], check=True)
    print("Dependency installation finished. If imports still fail, restart the runtime and rerun from cell 2.")
else:
    print("Required non-GUI dependencies already match the official versions.")
```

```python
import os
import sys
import importlib
import traceback

os.chdir(REPO_DIR)
os.environ["PYTHONPATH"] = str(REPO_DIR)
if str(REPO_DIR) not in sys.path:
    sys.path.insert(0, str(REPO_DIR))

IMPORT_CHECKS = {
    "torch": "torch / PyTorch",
    "transformers": "transformers==4.54.0",
    "datasets": "datasets==4.0.0",
    "accelerate": "accelerate==1.10.1",
    "miditoolkit": "miditoolkit==1.0.1",
    "huggingface_hub": "huggingface_hub==0.35.3",
    "src.model.pianoformer": "official PianoT5Gemma model module",
    "src.model.generate": "official generation module",
    "src.utils.download_model": "official checkpoint download utility",
}

failed = []
for module_name, package_hint in IMPORT_CHECKS.items():
    try:
        importlib.import_module(module_name)
        print(f"OK: {module_name}")
    except Exception as exc:
        failed.append((module_name, package_hint, exc))
        print(f"FAILED: {module_name} ({package_hint})")
        traceback.print_exception(type(exc), exc, exc.__traceback__)

if failed:
    hints = "; ".join([f"{module}: install/check {hint}" for module, hint, _ in failed])
    raise RuntimeError("Import check failed: " + hints)

from src.model.pianoformer import PianoT5Gemma
from src.model.generate import batch_performance_render, map_midi
from src.utils.download_model import download_hf
from miditoolkit import MidiFile

print("Official classes/functions ready:")
print("- PianoT5Gemma")
print("- batch_performance_render")
print("- map_midi")
print("- download_hf")
print("- MidiFile")
```

```python
# Official README states the fine-tuned rendering model is about 270 MB.
# Official repository utility downloads repo_id="yhj137/pianist-transformer-rendering" into models/sft.
CHECKPOINT_REPO_ID = "yhj137/pianist-transformer-rendering"
REQUIRED_CHECKPOINT_FILES = [
    "generation_config.json",
    "config.json",
    "model.safetensors",
]

CHECKPOINT_DIR.mkdir(parents=True, exist_ok=True)
missing_checkpoint_files = [name for name in REQUIRED_CHECKPOINT_FILES if not (CHECKPOINT_DIR / name).exists()]

if missing_checkpoint_files:
    print("Missing checkpoint files:", missing_checkpoint_files)
    print(f"Downloading official checkpoint from Hugging Face: {CHECKPOINT_REPO_ID}")
    download_hf(CHECKPOINT_REPO_ID, str(CHECKPOINT_DIR))
else:
    print("Checkpoint files already exist; skipping download.")

missing_after_download = []
for name in REQUIRED_CHECKPOINT_FILES:
    path = CHECKPOINT_DIR / name
    if not path.exists():
        missing_after_download.append(name)
    else:
        print(f"{name}: {path.stat().st_size / (1024 * 1024):.2f} MB at {path}")

if missing_after_download:
    raise FileNotFoundError("Checkpoint download did not produce required files: " + ", ".join(missing_after_download))

checkpoint_total_mb = sum(p.stat().st_size for p in CHECKPOINT_DIR.rglob("*") if p.is_file()) / (1024 * 1024)
print(f"Checkpoint directory total size: {checkpoint_total_mb:.2f} MB")
```

```python
midi_files = sorted(INPUT_MIDI_DIR.glob("*.mid"), key=lambda p: (p.stat().st_size, p.name))
if not midi_files:
    raise FileNotFoundError(f"No example MIDI files found under {INPUT_MIDI_DIR}")

print("Official example score MIDI files:")
for path in midi_files:
    print(f"- {path.name}: {path.stat().st_size} bytes")

# Choose the smallest bundled score MIDI as the simplest smoke-test input.
INPUT_MIDI_PATH = midi_files[0]
print("Selected input MIDI:", INPUT_MIDI_PATH)
print("Selected input size:", INPUT_MIDI_PATH.stat().st_size, "bytes")
```

```python
import torch

# Read official inference files and assert the flow we are reusing.
# We do not copy or modify source code; this cell imports official classes/functions and calls them directly.
official_inference_path = REPO_DIR / "src" / "inference" / "inference.py"
official_generate_path = REPO_DIR / "src" / "model" / "generate.py"

inference_text = official_inference_path.read_text(encoding="utf-8")
generate_text = official_generate_path.read_text(encoding="utf-8")

required_snippets = [
    "PianoT5Gemma.from_pretrained",
    "batch_performance_render",
    "map_midi",
    "data/midis/testset/score",
]
for snippet in required_snippets:
    if snippet not in inference_text:
        raise RuntimeError(f"Official inference.py no longer contains expected snippet: {snippet}")

if "def batch_performance_render" not in generate_text or "def map_midi" not in generate_text:
    raise RuntimeError("Official generate.py no longer exposes expected generation helpers.")

print("Official inference flow verified from source files.")
print("Official inference.py path:", official_inference_path)
print("Official generate.py path:", official_generate_path)

DEVICE = torch.device("cuda")

# Prefer bf16 only when the assigned CUDA device reports bf16 support.
# On many free Colab T4 runtimes, float16 is the safer choice.
if hasattr(torch.cuda, "is_bf16_supported") and torch.cuda.is_bf16_supported():
    MODEL_DTYPE = torch.bfloat16
else:
    MODEL_DTYPE = torch.float16

print("Selected CUDA dtype:", MODEL_DTYPE)
print("Reason: use bfloat16 only when CUDA reports bf16 support; otherwise use float16.")

model = PianoT5Gemma.from_pretrained(str(CHECKPOINT_DIR), torch_dtype=MODEL_DTYPE)
model.to(DEVICE)
model.eval()

first_param = next(model.parameters())
print("Model parameter device:", first_param.device)
print("Model parameter dtype:", first_param.dtype)

if first_param.device.type != "cuda":
    raise RuntimeError("Model is not on CUDA. Stop and inspect model.to('cuda') behavior.")

if first_param.dtype not in (torch.float16, torch.bfloat16, torch.float32):
    raise RuntimeError(f"Unexpected model dtype: {first_param.dtype}")

print("CUDA model load verified.")
```

```python
import time
import traceback

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print("Input MIDI:", INPUT_MIDI_PATH)
print("Output MIDI:", OUTPUT_MIDI_PATH)
print("Inference device:", DEVICE)
print("Inference dtype:", MODEL_DTYPE)

try:
    score_midi = MidiFile(str(INPUT_MIDI_PATH))
    start_time = time.perf_counter()
    with torch.inference_mode():
        rendered_midis, rendered_token_ids = batch_performance_render(
            model,
            [score_midi],
            temperature=1.0,
            top_p=0.95,
            device=str(DEVICE),
        )
        rendered_midi = map_midi(score_midi, rendered_midis[0])
        rendered_midi.dump(str(OUTPUT_MIDI_PATH))
    elapsed = time.perf_counter() - start_time
    print(f"Inference completed in {elapsed:.2f} seconds")
    print("Generated token count:", len(rendered_token_ids[0]) if rendered_token_ids else 0)
except Exception as exc:
    print("Inference failed. Full traceback follows:")
    traceback.print_exception(type(exc), exc, exc.__traceback__)
    raise
```

```python
if not OUTPUT_MIDI_PATH.exists():
    raise FileNotFoundError(f"Output MIDI does not exist: {OUTPUT_MIDI_PATH}")

output_size = OUTPUT_MIDI_PATH.stat().st_size
print("Output MIDI absolute path:", OUTPUT_MIDI_PATH.resolve())
print("Output MIDI size:", output_size, "bytes")

if output_size <= 0:
    raise RuntimeError("Output MIDI exists but is empty.")

output_midi = MidiFile(str(OUTPUT_MIDI_PATH))
note_count = sum(len(inst.notes) for inst in output_midi.instruments)
cc64_count = sum(1 for inst in output_midi.instruments for cc in inst.control_changes if cc.number == 64)
print("Instrument count:", len(output_midi.instruments))
print("Note count:", note_count)
print("CC64 sustain pedal event count:", cc64_count)

if note_count == 0:
    raise RuntimeError("Output MIDI contains no notes.")
```

```python
from google.colab import files

if not OUTPUT_MIDI_PATH.exists() or OUTPUT_MIDI_PATH.stat().st_size <= 0:
    raise RuntimeError("No non-empty output MIDI is available for download.")

files.download(str(OUTPUT_MIDI_PATH))
```

## Runtime Restart and Troubleshooting Notes

If the Colab runtime restarts, rerun these cells in order:

1. Runtime/GPU check.
2. Path definitions.
3. Clone repository, or reuse it if still present.
4. Dependency installation/import check.
5. Checkpoint download or restore.
6. MIDI selection.
7. Model load and inference.
8. Output verification and download.

Optional Google Drive checkpoint persistence:

- Minimal smoke test does not require Drive.
- After a successful checkpoint download, you may copy `models/sft/` to Drive.
- On a later runtime, copy it back into `/content/PianistTransformer/models/sft/` before loading the model.
- Prefer running inference from `/content`, not directly from mounted Drive, to avoid slow or quota-sensitive Drive I/O.

If free Colab GPU is not assigned:

- Confirm `Runtime > Change runtime type > GPU`.
- Reconnect and rerun the GPU check cell.
- Try later if Colab says no GPU is available.
- Do not continue with this notebook on CPU; it is intentionally GPU-first.

If CUDA OOM occurs:

- Restart the runtime and run only this notebook.
- Keep batch size at one MIDI file.
- Use the smallest bundled example MIDI first.
- Prefer `float16` unless the GPU supports `bfloat16`.
- If OOM persists, use a shorter MIDI or fall back to official CPU inference for diagnosis.
