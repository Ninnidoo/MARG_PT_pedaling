# Pianist Transformer Baseline Experiment Notebook

Purpose: run repeated baseline inference experiments with the official Pianist Transformer pretrained rendering checkpoint, using arbitrary score MIDI inputs and saving output MIDI plus structured metadata.

This notebook is for Google Colab GPU. It does not modify third-party source code and does not train Pianist Transformer.

## Cell Groups

Run once per new Colab session:

1. Runtime/GPU check.
2. Storage configuration and optional Google Drive mount.
3. Official repository clone/reuse and analyzed commit checkout verification.
4. Dependency setup.
5. Official module import, checkpoint preparation, and model load to CUDA.

Run repeatedly for each experiment input:

1. Select/upload an input MIDI.
2. Set reproducibility and generation parameters.
3. Run `N` performance samples without reloading the model.
4. Analyze CC64 sustain-pedal events.
5. Save output MIDI, JSON metadata, and CSV pedal-event table.

Important constraints:

- Do not call the official `script/inference.sh` for GPU experiments; it follows the official CPU inference path.
- Reuse official classes/functions from `third_party/PianistTransformer`: `PianoT5Gemma`, `batch_performance_render`, and `map_midi`.
- Treat fast off-on pedal transitions as exploratory candidates only. They are not classification labels because the repedal definition is not finalized.
- Files under `/content` are temporary. Use Google Drive or download important outputs before ending the runtime.

```python
# Session Setup 5: imports, checkpoint preparation, and one-time CUDA model load
import os
import sys
import json
import time
import random
import traceback
import shutil
from datetime import datetime, timezone
from pathlib import Path

import numpy as np
import pandas as pd
import torch
from miditoolkit import MidiFile

os.environ["PYTHONPATH"] = str(REPO_DIR)
if str(REPO_DIR) not in sys.path:
    sys.path.insert(0, str(REPO_DIR))

from src.model.pianoformer import PianoT5Gemma
from src.model.generate import batch_performance_render, map_midi
from src.utils.download_model import download_hf

CHECKPOINT_REPO_ID = "yhj137/pianist-transformer-rendering"
CHECKPOINT_DIR = CHECKPOINTS_DIR / "pianist-transformer-rendering"
REQUIRED_CHECKPOINT_FILES = ["generation_config.json", "config.json", "model.safetensors"]

CHECKPOINT_DIR.mkdir(parents=True, exist_ok=True)
missing_checkpoint_files = [name for name in REQUIRED_CHECKPOINT_FILES if not (CHECKPOINT_DIR / name).exists()]

if missing_checkpoint_files:
    print("Missing checkpoint files:", missing_checkpoint_files)
    print("Downloading official checkpoint. README notes this is about 270 MB.")
    download_hf(CHECKPOINT_REPO_ID, str(CHECKPOINT_DIR))
else:
    print("Checkpoint already exists; skipping download:", CHECKPOINT_DIR)

missing_after_download = [name for name in REQUIRED_CHECKPOINT_FILES if not (CHECKPOINT_DIR / name).exists()]
if missing_after_download:
    raise FileNotFoundError("Required checkpoint files missing: " + ", ".join(missing_after_download))

checkpoint_file_sizes = {name: (CHECKPOINT_DIR / name).stat().st_size for name in REQUIRED_CHECKPOINT_FILES}
print("Checkpoint files:", checkpoint_file_sizes)

DEVICE = torch.device("cuda")
if hasattr(torch.cuda, "is_bf16_supported") and torch.cuda.is_bf16_supported():
    MODEL_DTYPE = torch.bfloat16
else:
    MODEL_DTYPE = torch.float16
print("Selected model dtype:", MODEL_DTYPE)

model = PianoT5Gemma.from_pretrained(str(CHECKPOINT_DIR), torch_dtype=MODEL_DTYPE)
model.to(DEVICE)
model.eval()

first_param = next(model.parameters())
print("Model parameter device:", first_param.device)
print("Model parameter dtype:", first_param.dtype)
if first_param.device.type != "cuda":
    raise RuntimeError("Model is not on CUDA. Stop and inspect setup.")

print("Model is loaded once and ready for repeated experiments.")
```

## Repeated Experiment Cells

Run the cells below for each score MIDI you want to render. You can change input MIDI, seed, temperature, top-p, and number of samples without reloading the model.

The original input MIDI is never modified. Uploaded files are copied into the configured `inputs/` storage folder.

```python
# Experiment 1: choose or upload input MIDI
from google.colab import files

INPUT_SOURCE = "upload"  # Options: "upload" or "drive"
DRIVE_INPUT_INDEX = 0      # Used only when INPUT_SOURCE == "drive".

if INPUT_SOURCE == "upload":
    print("Upload one score MIDI file.")
    uploaded = files.upload()
    if not uploaded:
        raise RuntimeError("No file was uploaded.")
    uploaded_name = next(iter(uploaded.keys()))
    if not uploaded_name.lower().endswith((".mid", ".midi")):
        raise ValueError(f"Uploaded file does not look like MIDI: {uploaded_name}")
    input_midi_path = INPUTS_DIR / Path(uploaded_name).name
    with open(input_midi_path, "wb") as f:
        f.write(uploaded[uploaded_name])
    print("Uploaded input copied to:", input_midi_path)
elif INPUT_SOURCE == "drive":
    drive_midis = sorted(list(INPUTS_DIR.glob("*.mid")) + list(INPUTS_DIR.glob("*.midi")))
    if not drive_midis:
        raise FileNotFoundError(f"No MIDI files found in {INPUTS_DIR}")
    print("Available Drive input MIDI files:")
    for idx, path in enumerate(drive_midis):
        print(f"[{idx}] {path.name} ({path.stat().st_size} bytes)")
    input_midi_path = drive_midis[DRIVE_INPUT_INDEX]
else:
    raise ValueError("INPUT_SOURCE must be 'upload' or 'drive'.")

if not input_midi_path.exists() or input_midi_path.stat().st_size <= 0:
    raise FileNotFoundError(f"Input MIDI is missing or empty: {input_midi_path}")

input_midi = MidiFile(str(input_midi_path))
input_note_count = sum(len(inst.notes) for inst in input_midi.instruments)
input_cc64_count = sum(1 for inst in input_midi.instruments for cc in inst.control_changes if cc.number == 64)
input_total_cc_count = sum(len(inst.control_changes) for inst in input_midi.instruments)
input_instrument_count = len(input_midi.instruments)
input_tempo_count = len(input_midi.tempo_changes)
input_time_signature_count = len(input_midi.time_signature_changes)
input_max_tick = max(
    [0]
    + [note.end for inst in input_midi.instruments for note in inst.notes]
    + [cc.time for inst in input_midi.instruments for cc in inst.control_changes]
)

velocities = [note.velocity for inst in input_midi.instruments for note in inst.notes]
velocity_summary = {
    "min": int(min(velocities)) if velocities else None,
    "mean": float(np.mean(velocities)) if velocities else None,
    "max": int(max(velocities)) if velocities else None,
}

input_stats = {
    "path": str(input_midi_path),
    "size_bytes": input_midi_path.stat().st_size,
    "instrument_count": input_instrument_count,
    "note_count": input_note_count,
    "cc64_count": input_cc64_count,
    "total_control_change_count": input_total_cc_count,
    "tempo_change_count": input_tempo_count,
    "time_signature_change_count": input_time_signature_count,
    "max_tick": int(input_max_tick),
    "velocity_summary": velocity_summary,
}

print(json.dumps(input_stats, indent=2, ensure_ascii=False))
if input_note_count == 0:
    raise RuntimeError("Input MIDI has no notes; choose a score-like MIDI with piano notes.")
if input_instrument_count > 2:
    print("WARNING: This MIDI has multiple instruments. Inspect whether it is score-like enough for the baseline renderer.")
print("Original input MIDI was read for statistics only and was not modified.")
```

```python
# Experiment 2: reproducibility and generation parameters
SEED = 20260710
TEMPERATURE = 1.0
TOP_P = 0.95
NUM_SAMPLES = 1
EXPERIMENT_NAME = "baseline_pedal_probe"
FAST_OFF_ON_THRESHOLD_TICKS = 120  # exploratory candidate threshold, not a classification definition

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

# Deterministic flags can reduce nondeterminism but do not guarantee identical sampling across all CUDA kernels.
torch.backends.cudnn.benchmark = False
try:
    torch.use_deterministic_algorithms(False)
except Exception as exc:
    print("Could not set deterministic algorithms flag:", exc)

experiment_id = f"{datetime.now(timezone.utc).strftime('%Y%m%dT%H%M%SZ')}_{Path(input_midi_path).stem}"
print("Experiment ID:", experiment_id)
print("Seed:", SEED)
print("Temperature:", TEMPERATURE)
print("Top-p:", TOP_P)
print("Number of samples:", NUM_SAMPLES)
print("Repository commit:", ANALYZED_COMMIT)
print("Checkpoint repo ID:", CHECKPOINT_REPO_ID)
print("Fast off-on threshold ticks:", FAST_OFF_ON_THRESHOLD_TICKS, "(exploratory candidate only)")
```

```python
# Experiment 3: repeated inference without reloading the model
import re

def safe_float_for_filename(value):
    return str(value).replace(".", "p").replace("-", "m")

def safe_stem(value):
    return re.sub(r"[^A-Za-z0-9_.-]+", "_", value).strip("_") or "input"

run_records = []
input_stem = safe_stem(Path(input_midi_path).stem)
temp_tag = safe_float_for_filename(TEMPERATURE)
top_p_tag = safe_float_for_filename(TOP_P)

for run_index in range(1, NUM_SAMPLES + 1):
    sample_seed = SEED + run_index - 1
    random.seed(sample_seed)
    np.random.seed(sample_seed)
    torch.manual_seed(sample_seed)
    torch.cuda.manual_seed_all(sample_seed)

    output_filename = f"{input_stem}_seed{sample_seed}_temp{temp_tag}_top{top_p_tag}_run{run_index:02d}.mid"
    output_midi_path = OUTPUT_MIDI_DIR / output_filename

    print(f"Running sample {run_index}/{NUM_SAMPLES}: {output_filename}")
    score_midi = MidiFile(str(input_midi_path))
    started = time.perf_counter()
    try:
        with torch.inference_mode():
            rendered_midis, rendered_token_ids = batch_performance_render(
                model,
                [score_midi],
                temperature=TEMPERATURE,
                top_p=TOP_P,
                device=str(DEVICE),
            )
            rendered_midi = map_midi(score_midi, rendered_midis[0])
            rendered_midi.dump(str(output_midi_path))
    except Exception as exc:
        print("Inference failed. Traceback follows:")
        traceback.print_exception(type(exc), exc, exc.__traceback__)
        raise
    elapsed_seconds = time.perf_counter() - started
    token_count = len(rendered_token_ids[0]) if rendered_token_ids else 0
    output_size = output_midi_path.stat().st_size if output_midi_path.exists() else 0

    record = {
        "run_index": run_index,
        "sample_seed": sample_seed,
        "output_midi_path": str(output_midi_path),
        "output_filename": output_filename,
        "inference_time_seconds": elapsed_seconds,
        "generated_token_count": token_count,
        "output_size_bytes": output_size,
    }
    run_records.append(record)
    print(json.dumps(record, indent=2, ensure_ascii=False))

print("Completed samples:", len(run_records))
```

```python
# Experiment 4: pedal analysis for each output MIDI
from statistics import mean, median

PEDAL_ON_VALUE = 64

# Repedal definition is not finalized.
# fast_off_on_candidate_count is exploratory only and must not be used as a classification label.
def analyze_pedal_events(midi_path, fast_threshold_ticks=120):
    midi = MidiFile(str(midi_path))
    notes = [note for inst in midi.instruments for note in inst.notes]
    cc64_events = sorted(
        [cc for inst in midi.instruments for cc in inst.control_changes if cc.number == 64],
        key=lambda cc: (cc.time, cc.value),
    )

    event_rows = []
    for idx, cc in enumerate(cc64_events):
        event_rows.append({
            "output_midi_path": str(midi_path),
            "event_index": idx,
            "time": int(cc.time),
            "value": int(cc.value),
            "pedal_state": "on" if cc.value >= PEDAL_ON_VALUE else "off",
        })

    pedal_on_count = sum(1 for cc in cc64_events if cc.value >= PEDAL_ON_VALUE)
    pedal_off_count = sum(1 for cc in cc64_events if cc.value < PEDAL_ON_VALUE)

    intervals = []
    active_start = None
    for cc in cc64_events:
        is_on = cc.value >= PEDAL_ON_VALUE
        if is_on and active_start is None:
            active_start = int(cc.time)
        elif (not is_on) and active_start is not None:
            end = int(cc.time)
            intervals.append({
                "start_time": active_start,
                "end_time": end,
                "duration_ticks": max(0, end - active_start),
            })
            active_start = None
    if active_start is not None:
        intervals.append({
            "start_time": active_start,
            "end_time": None,
            "duration_ticks": None,
        })

    closed_durations = [item["duration_ticks"] for item in intervals if item["duration_ticks"] is not None]

    fast_candidates = 0
    for prev_cc, next_cc in zip(cc64_events, cc64_events[1:]):
        if prev_cc.value < PEDAL_ON_VALUE and next_cc.value >= PEDAL_ON_VALUE:
            if int(next_cc.time) - int(prev_cc.time) <= fast_threshold_ticks:
                fast_candidates += 1

    summary = {
        "output_midi_path": str(midi_path),
        "note_count": len(notes),
        "cc64_event_count": len(cc64_events),
        "pedal_on_event_count": pedal_on_count,
        "pedal_off_event_count": pedal_off_count,
        "pedal_on_interval_count": len(intervals),
        "pedal_on_duration_mean_ticks": float(mean(closed_durations)) if closed_durations else None,
        "pedal_on_duration_median_ticks": float(median(closed_durations)) if closed_durations else None,
        "pedal_on_duration_max_ticks": int(max(closed_durations)) if closed_durations else None,
        "fast_off_on_candidate_count": fast_candidates,
        "fast_off_on_threshold_ticks": fast_threshold_ticks,
        "candidate_note": "exploratory only; not a repedal classification label",
        "intervals": intervals,
    }
    return summary, event_rows

pedal_summaries = []
pedal_event_rows = []
for record in run_records:
    summary, rows = analyze_pedal_events(record["output_midi_path"], FAST_OFF_ON_THRESHOLD_TICKS)
    pedal_summaries.append(summary)
    pedal_event_rows.extend(rows)
    record["pedal_summary"] = {k: v for k, v in summary.items() if k != "intervals"}
    record["pedal_on_intervals"] = summary["intervals"]

summary_df = pd.DataFrame([{k: v for k, v in item.items() if k != "intervals"} for item in pedal_summaries])
events_df = pd.DataFrame(pedal_event_rows)

print("Pedal summary table:")
display(summary_df)
print("CC64 event table:")
display(events_df)
print("Reminder: fast off-on transitions are exploratory candidates, not classification labels.")
```

```python
# Experiment 5: save JSON metadata and CSV pedal-event table
metadata_path = METADATA_DIR / f"{experiment_id}_metadata.json"
pedal_csv_path = METADATA_DIR / f"{experiment_id}_pedal_events.csv"

metadata = {
    "schema_version": "1.0",
    "experiment_id": experiment_id,
    "experiment_name": EXPERIMENT_NAME,
    "created_utc": datetime.now(timezone.utc).isoformat(),
    "runtime": {
        "environment": "Google Colab GPU",
        "python_version": sys.version,
        "torch_version": torch.__version__,
        "torch_cuda_version": torch.version.cuda,
        "cuda_available": torch.cuda.is_available(),
        "gpu_name": torch.cuda.get_device_name(0) if torch.cuda.is_available() else None,
        "device": str(DEVICE),
        "model_dtype": str(MODEL_DTYPE),
    },
    "storage": {
        "use_google_drive": USE_GOOGLE_DRIVE,
        "storage_root": str(STORAGE_ROOT),
        "inputs_dir": str(INPUTS_DIR),
        "output_midi_dir": str(OUTPUT_MIDI_DIR),
        "metadata_dir": str(METADATA_DIR),
        "checkpoints_dir": str(CHECKPOINTS_DIR),
        "content_is_temporary": not USE_GOOGLE_DRIVE,
    },
    "model": {
        "repository_url": REPO_URL,
        "repository_commit": ANALYZED_COMMIT,
        "checkpoint_repo_id": CHECKPOINT_REPO_ID,
        "checkpoint_dir": str(CHECKPOINT_DIR),
        "third_party_source_modified": False,
    },
    "generation_parameters": {
        "seed": SEED,
        "temperature": TEMPERATURE,
        "top_p": TOP_P,
        "num_samples": NUM_SAMPLES,
    },
    "input": input_stats,
    "outputs": run_records,
    "pedal_analysis": {
        "pedal_on_value_threshold": PEDAL_ON_VALUE,
        "fast_off_on_threshold_ticks": FAST_OFF_ON_THRESHOLD_TICKS,
        "fast_off_on_candidate_definition": "off event followed by on event within threshold ticks; exploratory only, not a classification label",
    },
}

with open(metadata_path, "w", encoding="utf-8") as f:
    json.dump(metadata, f, indent=2, ensure_ascii=False)

events_df.to_csv(pedal_csv_path, index=False)

print("Saved metadata JSON:", metadata_path, metadata_path.stat().st_size, "bytes")
print("Saved pedal-event CSV:", pedal_csv_path, pedal_csv_path.stat().st_size, "bytes")
print("Saved output MIDI files:")
for record in run_records:
    path = Path(record["output_midi_path"])
    print("-", path, path.stat().st_size, "bytes")

if not USE_GOOGLE_DRIVE:
    from google.colab import files
    print("WARNING: Drive is disabled; offering downloads for temporary /content outputs.")
    files.download(str(metadata_path))
    files.download(str(pedal_csv_path))
    for record in run_records:
        files.download(record["output_midi_path"])
else:
    print("Google Drive storage is enabled; files are already written under:", STORAGE_ROOT)
```

## Metadata Schema

- `schema_version`
- `experiment_id`
- `experiment_name`
- `created_utc`
- `runtime`: Python, PyTorch, CUDA, GPU, device, dtype
- `storage`: Drive/content paths and persistence status
- `model`: official repo URL, analyzed commit, checkpoint repo ID, source modification flag
- `generation_parameters`: seed, temperature, top-p, sample count
- `input`: input MIDI path, size, note count, instrument count, CC statistics, score-like MIDI statistics
- `outputs`: per-sample output path, seed, runtime, generated token count, output size, pedal summary, pedal intervals
- `pedal_analysis`: thresholds and warning that fast off-on candidates are exploratory only
